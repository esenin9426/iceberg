# MERGE INTO: upsert и применение CDC-пакета изменений

Типичный сценарий в батчевых пайплайнах: из источника (OLTP-базы, Kafka,
файла выгрузки) прилетает пачка изменений — часть строк новые (`insert`),
часть обновлены (`update`), часть помечены на удаление (`delete`). Нужно
применить всё это одной операцией к целевой таблице так, чтобы результат
был согласован (или применилось всё, или ничего).

Именно для этого в Iceberg (как и в обычном ANSI SQL) есть `MERGE INTO`. Это
одна атомарная операция — она создаёт ровно один новый снапшот целевой
таблицы.

````python
from pyspark.sql import SparkSession

# Настраиваем каталог Iceberg прямо в коде, а не через spark-defaults.conf.
# Это даёт тот же результат, но позволяет менять параметры каталога
# независимо от конфигурации образа и делает ноутбук самодостаточным.
spark = (
    SparkSession.builder
    .appName("Merge into upsert")
    .master("local[*]")
    .config("spark.sql.extensions", "org.apache.iceberg.spark.extensions.IcebergSparkSessionExtensions")
    .config("spark.sql.catalog.course", "org.apache.iceberg.spark.SparkCatalog")
    .config("spark.sql.catalog.course.type", "rest")
    .config("spark.sql.catalog.course.uri", "http://rest-catalog:8181")
    .config("spark.sql.catalog.course.io-impl", "org.apache.iceberg.aws.s3.S3FileIO")
    .config("spark.sql.catalog.course.warehouse", "s3://warehouse")
    .config("spark.sql.catalog.course.s3.endpoint", "http://minio:9000")
    .config("spark.sql.catalog.course.s3.path-style-access", "true")
    .getOrCreate()
)

print("Spark + Iceberg готовы к работе")
````

Ниже определена вспомогательная функция `read_metadata_table` для чтения
служебных metadata-таблиц Iceberg в обход REST-каталога этого стенда —
подробное объяснение, зачем она нужна, см. в `00_Setup_and_Basics.ipynb`.

````python
from pyspark.sql import DataFrame


def read_metadata_table(spark, table_identifier, metadata_type):
    """Читает metadata-таблицу Iceberg (history, snapshots, files, partitions, manifests, ...)
    в обход REST-каталога, который отклоняет составные namespace в URL
    (см. подробное объяснение в 00_Setup_and_Basics.ipynb)."""
    jvm = spark._jvm
    table = jvm.org.apache.iceberg.spark.Spark3Util.loadIcebergTable(spark._jsparkSession, table_identifier)
    mtype = jvm.org.apache.iceberg.MetadataTableType.valueOf(metadata_type.upper())
    jdf = jvm.org.apache.iceberg.spark.SparkTableUtil.loadMetadataTable(spark._jsparkSession, table, mtype)
    return DataFrame(jdf, spark)
````

````python
spark.sql("CREATE NAMESPACE IF NOT EXISTS course.test")

spark.sql("DROP TABLE IF EXISTS course.test.customers")
spark.sql("""
CREATE TABLE course.test.customers (
    id      BIGINT,
    name    STRING,
    city    STRING
)
USING iceberg
""")

spark.sql("""
INSERT INTO course.test.customers VALUES
    (1, 'Alice', 'Moscow'),
    (2, 'Bob',   'Berlin'),
    (3, 'Carol', 'London')
""")

spark.sql("SELECT * FROM course.test.customers ORDER BY id").show()
````

## Готовим пакет изменений

Источник изменений оформляем как обычный DataFrame с колонкой `op`:
- `U` — обновить строку (или вставить, если её ещё нет — "upsert");
- `D` — удалить строку;
- `I` — вставить новую строку.

На практике этот DataFrame чаще всего приходит из CDC-потока (Debezium,
Kafka Connect) или из ежедневной выгрузки правок из другой системы.

````python
changes = spark.createDataFrame(
    [
        (1, "Alice", "Dubai", "U"),   # Alice переехала -> обновление
        (2, None,    None,    "D"),   # Bob ушёл -> удаление
        (4, "Dylan", "Paris", "I"),   # новый клиент -> вставка
    ],
    ["id", "name", "city", "op"],
)
changes.createOrReplaceTempView("customers_changes")
changes.show()
````

## MERGE INTO

- `WHEN MATCHED AND op = 'D'` — если строка есть и помечена на удаление,
  удаляем;
- `WHEN MATCHED` (для оставшихся, то есть `op = 'U'`) — обновляем все
  колонки;
- `WHEN NOT MATCHED` — строки нет в целевой таблице, вставляем как новую.

````python
spark.sql("""
MERGE INTO course.test.customers t
USING customers_changes s
ON t.id = s.id
WHEN MATCHED AND s.op = 'D' THEN DELETE
WHEN MATCHED THEN UPDATE SET t.name = s.name, t.city = s.city
WHEN NOT MATCHED THEN INSERT (id, name, city) VALUES (s.id, s.name, s.city)
""")

spark.sql("SELECT * FROM course.test.customers ORDER BY id").show()
````

## Это одна операция — один снапшот

Несмотря на то, что `MERGE` затронул сразу вставку, обновление и удаление,
в истории таблицы это ровно один новый снапшот с `operation = overwrite`.

````python
read_metadata_table(spark, "course.test.customers", "snapshots") \
    .select("committed_at", "snapshot_id", "operation") \
    .show(truncate=False)
````

## Частый вариант: чистый upsert без удалений

Если источник никогда не присылает удаления (например, это просто
дедуплицированный дневной снепшот измерения), `MERGE` можно упростить до
двух веток — обновление совпавших и вставка новых:

```sql
MERGE INTO course.test.customers t
USING customers_changes s
ON t.id = s.id
WHEN MATCHED THEN UPDATE SET *
WHEN NOT MATCHED THEN INSERT *
```

`UPDATE SET *` и `INSERT *` — это сокращённая форма: колонки сопоставляются
по имени между целевой таблицей и источником.
