# Iceberg + Spark: подключение и первые шаги

Этот ноутбук объясняет, из чего состоит окружение проекта, и показывает базовые
операции с таблицей Apache Iceberg через Spark: создание, вставку, выборку,
обновление и удаление данных.

## Из чего состоит стенд (см. `docker-compose.yml`)

| Сервис          | Роль                                                                                     |
|-----------------|-------------------------------------------------------------------------------------------|
| `minio`         | S3-совместимое хранилище. Здесь физически лежат data-файлы (parquet) и metadata-файлы (json/avro) таблиц. |
| `rest-catalog`  | Iceberg REST Catalog (`apache/iceberg-rest-fixture`). Хранит указатель на *текущий* `metadata.json` каждой таблицы и обеспечивает атомарную смену снапшотов (commit). |
| `spark_notebook`| Jupyter + PySpark с докаченными jar-файлами Iceberg, Hadoop-AWS и AWS SDK (см. `spark_notebook/Dockerfile`). Это "движок", который читает/пишет файлы в MinIO и общается с REST-каталогом по HTTP. |

Ключевая идея Iceberg: **Spark никогда не ходит напрямую в Hive Metastore или в
файловую систему, чтобы понять "что такое таблица сейчас"**. Он спрашивает у
каталога (здесь — REST-каталог) путь к актуальному `metadata.json`, а дальше
уже сам читает manifest list → manifest files → data files из S3.

## Настройка Spark-сессии

Каталог с именем `course` подключается через набор `spark.sql.catalog.course.*`
опций. Это можно сделать двумя способами:

1. Через `spark-defaults.conf` в образе (см. `spark_notebook/spark-defaults copy.conf`) — тогда настройки едины для всех ноутбуков.
2. Прямо в коде через `SparkSession.builder.config(...)` — как сделано ниже. Это удобно, когда конфиг образа урезан (например, для разбора "сырых" файлов метаданных, как в `Table_structure.ipynb`) или когда нужно быстро поднять сессию без пересборки контейнера.

```python
from pyspark.sql import SparkSession

# Настраиваем каталог Iceberg прямо в коде, а не через spark-defaults.conf.
# Это даёт тот же результат, но позволяет менять параметры каталога
# независимо от конфигурации образа и делает ноутбук самодостаточным.
spark = (
    SparkSession.builder
    .appName("Iceberg basics")
    .master("local[*]")
    .config("spark.sql.extensions", "org.apache.iceberg.spark.extensions.IcebergSparkSessionExtensions")
    .config("spark.sql.catalog.course", "org.apache.iceberg.spark.SparkCatalog")
    .config("spark.sql.catalog.course.type", "rest")
    .config("spark.sql.catalog.course.uri", "http://rest-catalog:8181")
    .config("spark.sql.catalog.course.io-impl", "org.apache.iceberg.aws.s3.S3FileIO")
    .config("spark.sql.catalog.course.warehouse", "s3://warehouse")
    .config("spark.sql.catalog.course.s3.endpoint", "http://minio:9000")
    .config("spark.sql.catalog.course.s3.path-style-access", "true")
    .getOrCreate()
)

print("Spark + Iceberg готовы к работе")
```

## Namespace и таблица

В Iceberg таблица всегда живёт в неймспейсе (аналог схемы/базы данных):
`<каталог>.<namespace>.<таблица>`, например `course.test.users`. Именно
поэтому в старых ноутбуках проекта встречается ошибка
`REQUIRES_SINGLE_PART_NAMESPACE` — она возникает, если случайно писать в
`spark_catalog` (каталог по умолчанию) с многосоставным именем, а не в
`course`.

```python
spark.sql("CREATE NAMESPACE IF NOT EXISTS course.test")
spark.sql("SHOW NAMESPACES IN course").show()
```

```python
spark.sql("""
CREATE TABLE IF NOT EXISTS course.test.basics (
    id      BIGINT,
    name    STRING,
    city    STRING
)
USING iceberg
""")

spark.sql("SHOW TABLES IN course.test").show()
```

## Запись данных: SQL и DataFrame API

Iceberg-таблицу можно наполнять и обычным `INSERT INTO` в SQL, и через
`DataFrameWriterV2` (`writeTo`). Оба способа порождают новый снапшот таблицы.

```python
spark.sql("""
INSERT INTO course.test.basics VALUES
    (1, 'Alice', 'Moscow'),
    (2, 'Bob',   'Berlin')
""")
```

```python
from pyspark.sql import Row

new_rows = spark.createDataFrame([Row(id=3, name="Carol", city="London")])
new_rows.writeTo("course.test.basics").append()

spark.sql("SELECT * FROM course.test.basics ORDER BY id").show()
```

## Обновление и удаление

Iceberg v2 поддерживает `UPDATE` и `DELETE` на уровне SQL из коробки — под
капотом это delete-файлы (position/equality deletes), которые Spark
применяет "на лету" при чтении, не переписывая сразу все data-файлы.

```python
spark.sql("UPDATE course.test.basics SET city = 'Dubai' WHERE id = 1")
spark.sql("DELETE FROM course.test.basics WHERE id = 2")

spark.sql("SELECT * FROM course.test.basics ORDER BY id").show()
```

## Служебная информация о таблице

Полезные команды, чтобы посмотреть текущее состояние таблицы без похода во
внутренние metadata-таблицы:

```python
spark.sql("DESCRIBE TABLE course.test.basics").show(truncate=False)
```

```python
spark.sql("SHOW CREATE TABLE course.test.basics").show(truncate=False)
```
